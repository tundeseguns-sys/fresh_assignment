```python
# write a for loop program that print all prime numbers from the given list.
list_1 = [12,21,31,7,9,3,41,50,32,39,67]
for i in list_1:
    for j in range(2, i):
        if i % j == 0:
            break
    else:
        print(f'{i} is a prime number')
```

```python
list_1 = [12,21,31,7,9,3,41,50,32,39,67]

for i in list_1:
    for j in range(2, int(i ** 0.5)):
        if i % j == 0:
            break
    else:
      print(f'{i} is a prime number')
```

```python
# The "Perfect" Balance
list_1 = [12, 21, 31, 7, 9, 3, 41, 50, 32, 39, 67]

for i in list_1:
    if i < 2: continue

    # Speed of square root + Safety of "+1"
    for j in range(2, int(i ** 0.5) + 1):
        if i % j == 0:
            break
    else:
        print(f'{i} is a prime number')
```

```python
# recursion - when a function calls itself # 5! 5*4*3*2*1

# calculate factorial using recursion
def factorial(n):
    '''Calculate factorial using recursion'''
    if n == 0 or n == 1:
        return 1
    # recursion case
    return n * factorial(n-1)

print(factorial(5))
```

```python
def factorial(n):
    # 1. BASE CASE: Factorial of 1 is 1. Stop here.
    if n == 1:
        return 1

    # 2. RECURSIVE STEP: n * (factorial of n-1)
    else:
        return n * factorial(n - 1)

print(factorial(5))  # Output: 120
```

How Python Executes This (The "Stack")
When you run factorial(5), Python pauses that function to go calculate factorial(4). It builds a "stack" of paused tasks waiting for answers.
- factorial(5) asks: "What is 5 * factorial(4)?" (Pauses)
- factorial(4) asks: "What is 4 * factorial(3)?" (Pauses)
- factorial(3) asks: "What is 3 * factorial(2)?" (Pauses)
- factorial(2) asks: "What is 2 * factorial(1)?" (Pauses)
- factorial(1) says: "I am 1!" (Base Case Reached!)

Now it unwinds back up:
- factorial(2) hears "1". It calculates $2 \times 1 = 2$.
- factorial(3) hears "2". It calculates $3 \times 2 = 6$.
- factorial(4) hears "6". It calculates $4 \times 6 = 24$.
- factorial(5) hears "24". It calculates $5 \times 24 = 120$.
